# HANDS-ON EXERCISE 2 — Hospital Patient Data Cleaning

**Industry:** Healthcare

## Part A — Initial Inspection

The task requires `head()`, `shape`, `info()`, and `describe()`, followed by documentation of what we discover.


```python
df_hospital_raw = pd.read_csv("hospital_patient_data_cleaning_practice.csv")

display(df_hospital_raw.head())

print("Shape:", df_hospital_raw.shape)

print("\nInfo:")
df_hospital_raw.info()

print("\nDescriptive statistics:")
display(df_hospital_raw.describe(include="all"))
```

### What was discovered

The supplied hospital dataset contains:

- Missing City values.
- Missing Age values.
- Missing Visit_Count values.
- Text values in `Visit_Count`, including values such as `seven`, `ten`, `six`, and `4 visits`.
- Exact duplicate rows.
- Inconsistent city labels such as `ABUJA`, `ibadan `, `lagos`, and `Port-Harcourt`.
- Records clearly marked as `TEST` with dummy/test patient information.

The exercise explicitly says to investigate before deleting problematic data, so the next cells quantify these issues first.


## Part B — Investigate Data Problems


```python
print("Missing values by column:")
display(df_hospital_raw.isna().sum())

print("Exact duplicate rows:", df_hospital_raw.duplicated().sum())

print("\nUnexpected Visit_Count values:")
visit_numeric_test = pd.to_numeric(df_hospital_raw["Visit_Count"], errors="coerce")
display(df_hospital_raw.loc[
    visit_numeric_test.isna() & df_hospital_raw["Visit_Count"].notna(),
    ["Patient_ID", "Patient_Name", "Visit_Count"]
].drop_duplicates().head(30))

print("\nPotential test/dummy records:")
display(df_hospital_raw[
    df_hospital_raw["City"].fillna("").astype(str).str.strip().str.upper().eq("TEST")
])

print("\nCity values that need normalization:")
display(df_hospital_raw["City"].value_counts(dropna=False).sort_index())
```

## Cleaning decisions

1. **Exact duplicate rows:** remove them because they contain the same complete record.
2. **Test/dummy records:** remove rows whose City is `TEST`; these are explicitly labelled as test/sample records rather than genuine patient activity.
3. **City:** replace missing values with `Unknown`, as required by the exercise.
4. **Inconsistent city spelling/capitalization:** normalize labels, including `ABUJA` → `Abuja`, `ibadan ` → `Ibadan`, `lagos` → `Lagos`, and `Port-Harcourt` → `Port Harcourt`.
5. **Visit_Count:** convert recoverable text values (`seven`, `ten`, `six`, `4 visits`) into numbers. Rows still missing Visit_Count are removed because visit count is required for patient classification and average visit calculations.
6. **Age:** missing Age values are retained because Age is not required for any of the requested calculations. Filling them with an invented value would add an unsupported assumption.


```python
df_hospital = df_hospital_raw.drop_duplicates().copy()

# Remove clearly labelled test/dummy records.
is_test_record = df_hospital["City"].fillna("").astype(str).str.strip().str.upper().eq("TEST")
test_records_removed = is_test_record.sum()
df_hospital = df_hospital.loc[~is_test_record].copy()

# Fill missing cities as required.
df_hospital["City"] = df_hospital["City"].fillna("Unknown")

# Normalize inconsistent city labels.
df_hospital["City"] = df_hospital["City"].astype(str).str.strip().str.title()
df_hospital["City"] = df_hospital["City"].replace({
    "Port-Harcourt": "Port Harcourt"
})

# Clean Visit_Count.
visit_clean = (
    df_hospital["Visit_Count"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({"seven": "7", "ten": "10", "six": "6"})
    .str.replace(" visits", "", regex=False)
)

df_hospital["Visit_Count"] = pd.to_numeric(visit_clean, errors="coerce")

# Rows with no usable visit count cannot be classified or included in visit averages.
invalid_visit_rows = df_hospital["Visit_Count"].isna().sum()
df_hospital = df_hospital.dropna(subset=["Visit_Count"]).copy()

print("Test/dummy records removed:", test_records_removed)
print("Rows with unusable Visit_Count removed:", invalid_visit_rows)
print("Records remaining after cleaning:", len(df_hospital))
print("\nRemaining missing values:")
display(df_hospital.isna().sum())
```

## Part C — Missing Values

The exercise specifically requires missing City values to become `Unknown`. The cleaning above applies and verifies that rule.


```python
print("Unknown City records after cleaning:", (df_hospital["City"] == "Unknown").sum())
print("Missing City values after cleaning:", df_hospital["City"].isna().sum())
```

## Part D — Patient Classification

Create `Patient_Category`:

- Visit count >= 5 → **Frequent Patient**
- Visit count < 5 → **Occasional Patient**


```python
df_hospital["Patient_Category"] = df_hospital["Visit_Count"].apply(
    lambda x: "Frequent Patient" if x >= 5 else "Occasional Patient"
)

display(df_hospital[["Patient_ID", "Visit_Count", "Patient_Category"]].head(10))
```

## Part E — Quick Analysis

### 1. Number of patients by city


```python
patients_by_city = df_hospital["City"].value_counts()
display(patients_by_city)
```

### 2. Number of frequent patients


```python
frequent_patients = (df_hospital["Patient_Category"] == "Frequent Patient").sum()
print("Frequent patients:", frequent_patients)
```

### 3. Average patient visit count


```python
average_visits = df_hospital["Visit_Count"].mean()
print(f"Average patient visit count: {average_visits:.2f}")
```

### 4. Number of records with missing information before cleaning


```python
hospital_missing_records_before = df_hospital_raw.isna().any(axis=1).sum()
print("Records with at least one missing value before cleaning:", hospital_missing_records_before)
```

### 5. Number of records remaining after cleaning


```python
print("Records remaining after cleaning:", len(df_hospital))
```

### Hospital analysis interpretation

After the documented cleaning steps, **2,953 records remain**. There are **1,986 Frequent Patients**, and the average visit count is approximately **6.52 visits**.

The city distribution should be interpreted after normalization, because otherwise capitalization and spelling differences would make the same city appear as separate groups.


## Exercise 2 — Challenge

Create a summary table with:

| Patient Category | Number of Patients | Average Visits |
|---|---:|---:|
| Frequent Patient | ? | ? |
| Occasional Patient | ? | ? |


```python
hospital_summary = (
    df_hospital.groupby("Patient_Category")["Visit_Count"]
    .agg(Number_of_Patients="count", Average_Visits="mean")
    .reindex(["Frequent Patient", "Occasional Patient"])
)

display(hospital_summary)
```

### Challenge interpretation

The Frequent Patient group contains **1,986 patients** and averages about **8.48 visits**, while the Occasional Patient group contains **967 patients** and averages about **2.51 visits**. This shows a clear difference in visit activity between the two groups and gives the hospital a useful starting point for investigating why some patients return much more often.


# HANDS-ON EXERCISE 3 — Banking Customer Transaction Analysis

**Industry:** Banking / Financial Services

## Part A — Load and Inspect


```python
df_banking_raw = pd.read_csv("banking_customer_transaction.csv")

display(df_banking_raw.head())

print("Dataset size:", df_banking_raw.shape)

print("\nColumn names:")
print(df_banking_raw.columns.tolist())

print("\nData types:")
df_banking_raw.info()

print("\nNumerical summary statistics:")
display(df_banking_raw.describe(include="all"))
```

## Part B — Investigate the Transaction Amount

We inspect missing values, unexpected text, and invalid records before cleaning.


```python
print("Missing values by column:")
display(df_banking_raw.isna().sum())

amount_numeric_test = pd.to_numeric(df_banking_raw["Amount"], errors="coerce")

print("\nUnexpected/non-numeric Amount values:")
display(df_banking_raw.loc[
    amount_numeric_test.isna() & df_banking_raw["Amount"].notna(),
    ["Customer ID", "Amount"]
].drop_duplicates().head(50))

print("\nExact duplicate rows:", df_banking_raw.duplicated().sum())
```

### Cleaning decisions

- Missing categorical values are represented as `Unknown` so the records are not unnecessarily discarded.
- Currency-formatted values such as `₦12,500` can be converted safely to numeric.
- `1O000` is treated as a recoverable typing error where the letter `O` is clearly being used in place of zero, so it is converted to `10000`.
- Placeholder/error values such as `Not Available`, `Error`, `Pending`, `Unknown`, and `##VALUE!` cannot be reliably inferred as transaction amounts, so their Amount becomes missing and those rows are removed from amount-based analysis.
- Repeated Customer IDs are **not** automatically removed because a banking customer can legitimately have multiple transactions. There are no exact duplicate rows.


```python
df_banking = df_banking_raw.copy()

# Fill missing categorical information without inventing a category.
for col in ["Customer Name", "City", "Transaction Type", "Account Status"]:
    df_banking[col] = df_banking[col].fillna("Unknown")

# Clean recoverable numeric formatting/typos.
amount_clean = (
    df_banking["Amount"]
    .astype(str)
    .str.replace("₦", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace("O", "0", regex=False)
)

df_banking["Amount"] = pd.to_numeric(amount_clean, errors="coerce")

invalid_amount_rows = df_banking["Amount"].isna().sum()
df_banking = df_banking.dropna(subset=["Amount"]).copy()

print("Rows with unusable transaction amounts removed:", invalid_amount_rows)
print("Records remaining after cleaning:", len(df_banking))
print("\nRemaining missing values:")
display(df_banking.isna().sum())
```

## Part C — Clean the Dataset

The cleaning above is intentionally documented: recoverable formatting errors are repaired, while values that cannot safely be reconstructed are excluded from calculations that require a numeric transaction amount.


## Part D — Create Transaction Classification

- Amount >= 100,000 → **Large Transaction**
- Amount < 100,000 → **Regular Transaction**


```python
df_banking["Transaction_Category"] = df_banking["Amount"].apply(
    lambda x: "Large Transaction" if x >= 100000 else "Regular Transaction"
)

display(df_banking[["Customer ID", "Amount", "Transaction_Category"]].head(10))
```

## Part E — Business Analysis

### 1. How many transactions occurred in each city?


```python
transactions_by_city = df_banking["City"].value_counts()
display(transactions_by_city)
```

### 2. What is the average transaction amount?


```python
average_transaction = df_banking["Amount"].mean()
print(f"Average transaction amount: {average_transaction:,.2f}")
```

### 3. How many large transactions occurred?


```python
large_transactions = (df_banking["Transaction_Category"] == "Large Transaction").sum()
print("Large transactions:", large_transactions)
```

### 4. Which transaction type appears most frequently?


```python
transaction_type_counts = df_banking["Transaction Type"].value_counts()
display(transaction_type_counts)
print("Most frequent transaction type:", transaction_type_counts.idxmax())
```

### 5. What is the average transaction amount by transaction type?


```python
average_by_transaction_type = (
    df_banking.groupby("Transaction Type")["Amount"]
    .mean()
    .sort_values(ascending=False)
)
display(average_by_transaction_type)
```

### Banking analysis interpretation

After cleaning, **2,814 transaction records** remain for amount-based analysis. The average transaction amount is approximately **45,076.66**, and there are **276 Large Transactions**.

`Transfer` is the most frequent transaction type in the cleaned data. The grouped averages show how typical transaction values differ across transaction types.


## Exercise 3 — Challenge

Use `groupby()` to determine the average transaction amount by city. Then identify the city with the highest average and explain what it could mean from a business-analysis perspective.


```python
average_by_city = (
    df_banking.groupby("City")["Amount"]
    .mean()
    .sort_values(ascending=False)
)

display(average_by_city)

highest_city = average_by_city.idxmax()
highest_city_average = average_by_city.max()

print(f"City with the highest average transaction amount: {highest_city}")
print(f"Average transaction amount in {highest_city}: {highest_city_average:,.2f}")
```

### Challenge interpretation

**Benin City** has the highest average transaction amount at approximately **52,154.37**. This could indicate that the transactions represented in the dataset from Benin City tend to be larger on average.

However, this is an **observation, not a final business decision**. The bank should investigate transaction volume, customer mix, account types, transaction types, and the distribution of unusually large transactions before concluding why the city has a higher average.


# HANDS-ON EXERCISE 4 — Energy Consumption Analysis

**Industry:** Energy / Utilities

## Part A — Understand the Data


```python
df_energy_raw = pd.read_csv("energy_consumption.csv")

display(df_energy_raw.head())

print("Number of records:", df_energy_raw.shape[0])
print("Number of columns:", df_energy_raw.shape[1])

print("\nData types:")
df_energy_raw.info()

print("\nSummary statistics:")
display(df_energy_raw.describe(include="all"))
```

## Part B — Data Cleaning

We investigate:

- Missing City values
- Missing consumption values
- Unexpected text values
- Unnecessary rows

The dataset contains exact duplicate rows, missing categorical values, and non-numeric consumption placeholders. Some malformed numeric strings are recoverable, such as `12O` and `5,2O0`.


```python
print("Missing values by column:")
display(df_energy_raw.isna().sum())

print("Exact duplicate rows:", df_energy_raw.duplicated().sum())

consumption_numeric_test = pd.to_numeric(df_energy_raw["Monthly Consumption"], errors="coerce")

print("\nUnexpected/non-numeric consumption values:")
display(df_energy_raw.loc[
    consumption_numeric_test.isna() & df_energy_raw["Monthly Consumption"].notna(),
    ["Customer ID", "Monthly Consumption"]
].drop_duplicates().head(50))
```

### Energy cleaning decisions

1. Remove exact duplicate rows.
2. Replace missing categorical information with `Unknown`.
3. Correct clearly recoverable numeric formatting such as commas and the letter `O` used in place of zero.
4. Convert remaining non-numeric placeholders to missing.
5. Remove records with no usable Monthly Consumption because the requested consumption averages and categories cannot be calculated for them.


```python
df_energy = df_energy_raw.drop_duplicates().copy()

for col in ["City", "Customer Type", "Account Status", "Energy Source"]:
    df_energy[col] = df_energy[col].fillna("Unknown")

consumption_clean = (
    df_energy["Monthly Consumption"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("O", "0", regex=False)
)

df_energy["Monthly Consumption"] = pd.to_numeric(consumption_clean, errors="coerce")

invalid_consumption_rows = df_energy["Monthly Consumption"].isna().sum()
df_energy = df_energy.dropna(subset=["Monthly Consumption"]).copy()

print("Rows with unusable consumption removed:", invalid_consumption_rows)
print("Records remaining after cleaning:", len(df_energy))
print("\nRemaining missing values:")
display(df_energy.isna().sum())
```

## Part C — Create Consumption Category

- Consumption >= 500 → **High Consumption**
- Consumption < 500 → **Normal Consumption**


```python
df_energy["Consumption_Category"] = df_energy["Monthly Consumption"].apply(
    lambda x: "High Consumption" if x >= 500 else "Normal Consumption"
)

display(df_energy[["Customer ID", "Monthly Consumption", "Consumption_Category"]].head(10))
```

## Part D — Analyze Consumption

### 1. Number of customers in each city


```python
customers_by_energy_city = df_energy["City"].value_counts()
display(customers_by_energy_city)
```

### 2. Average consumption


```python
average_consumption = df_energy["Monthly Consumption"].mean()
print(f"Average monthly consumption: {average_consumption:,.2f}")
```

### 3. Number of high-consumption customers


```python
high_consumption_count = (df_energy["Consumption_Category"] == "High Consumption").sum()
print("High-consumption customers:", high_consumption_count)
```

### 4. Average consumption by customer type


```python
average_consumption_by_type = (
    df_energy.groupby("Customer Type")["Monthly Consumption"]
    .mean()
    .sort_values(ascending=False)
)
display(average_consumption_by_type)
```

### 5. Number of customers using each energy source


```python
energy_source_counts = df_energy["Energy Source"].value_counts()
display(energy_source_counts)
```

### Energy analysis interpretation

After cleaning, **2,813 records** remain. Average monthly consumption is approximately **1,078.73**, and **1,284 customers** fall into the High Consumption category.

Industrial customers have the highest average consumption in the cleaned data, while Residential customers have the largest number of records. These are useful patterns for further investigation, but they are not by themselves a final business decision.


## Exercise 4 — Challenge

Create a summary showing:

- Customer Type
- Average Consumption
- Number of Customers

Then write a short recommendation for what the energy company should investigate next.


```python
energy_summary = (
    df_energy.groupby("Customer Type")["Monthly Consumption"]
    .agg(
        Average_Consumption="mean",
        Number_of_Customers="count"
    )
    .sort_values("Average_Consumption", ascending=False)
)

display(energy_summary)
```

### Challenge recommendation

The energy company should investigate **why Industrial customers have a much higher average monthly consumption than the other customer types**, including whether this is explained by customer size, operating hours, equipment, tariff structure, or the number of high-consumption accounts.

The company should also examine the unusually high-consumption records individually before making operational or pricing decisions. This follows the exercise's instruction to identify a useful pattern for further investigation rather than making a final business decision from a simple analysis.
